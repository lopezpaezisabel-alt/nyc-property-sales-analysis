## Data Profiling

The raw NYC property sales dataset contains 845,607 records and 29 columns.

The initial profiling identified several data quality issues:

1. 3,351 exact duplicate records
2. Missing values in several analytical fields, including 'TOTAL UNITS', 'RESIDENTIAL UNITS', 'COMMERCIAL UNITS', 'LAND SQUARE FEET', 'GROSS SQUARE FEET', and 'YEAR BUILT'
3. Incorrect data types where numeric fields were imported as text and will require type conversion before analysis
4. The 'EASE-MENT' column contains NaN
5. 'SALE PRICE' contains valid $0 ownership transfers that will require separate tratment from the cash-based sales analysis

These issues are to be resolved during the cleaning stages while preserving valid records and documenting the reasoning behind each of the transformations.
